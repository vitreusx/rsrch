```python
from utils import *
from plotly.subplots import make_subplots
from tqdm.auto import tqdm
import json
from loaders import *
```

```python
res_df, scalars = data_aug_loader()
res_b_df, scalars_b = baseline_loader()
```

```python
envs = res_df["env"].unique()

fig = make_subplots(
    rows=len(envs),
    row_titles=[*envs],
    cols=1,
)

for row, env in enumerate(envs, 1):
    df1 = res_df[res_df["env"] == env].copy()
    df1["tag"] = df1["ratio"].apply(lambda r: f"aug/{r}")
    df2 = res_b_df[res_b_df["env"] == env].copy()
    df2 = df2[df2["ratio"].isin(df1["ratio"].unique())]
    df2["tag"] = df2["ratio"].apply(lambda r: f"base/{r}")
    df = pd.concat([df1, df2])
    df = df.sort_values(by="tag")
    traces = [go.Box(x=df["tag"], y=df["score"], boxmean=True)]
    for trace in traces:
        fig.add_trace(trace, row=row, col=1)

fig
```
